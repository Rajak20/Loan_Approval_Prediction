```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df[" education"] = le.fit_transform(df[" education"])
```

```python
df[" self_employed"] = le.fit_transform(df[" self_employed"])
df[" loan_status"] = le.fit_transform(df[" loan_status"])
```

```python
df.head()
```

```python
# Calculate the correlation of all numerical/encoded columns with loan_status
# Drop 'loan_id' as it is just an identifier and won't contain meaningful correlation
correlation_matrix = df.drop(columns=['loan_id']).corr()
loan_status_corr = correlation_matrix[' loan_status'].sort_values(ascending=False)

print("Correlation of features with loan_status:")
print(loan_status_corr)
```

```python
# Plot the heatmap for the correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap (including loan_status)', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Identify features to compare with loan_status (excluding loan_id and loan_status itself)
features_to_plot = [col for col in df.columns if col.strip() not in ['loan_id', 'loan_status']]

# Set up a grid of subplots
num_features = len(features_to_plot)
cols = 3
rows = (num_features + cols - 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows))
axes = axes.flatten()

for i, col in enumerate(features_to_plot):
    # Plot histogram colored by loan_status (0 = Approved, 1 = Rejected)
    sns.histplot(data=df, x=col, hue=" loan_status", kde=True, ax=axes[i], multiple="stack", palette="Set2")
    axes[i].set_title(f'Distribution of {col.strip()} by Loan Status')
    axes[i].set_xlabel(col.strip())
    axes[i].set_ylabel('Count')

# Hide unused axes
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
x = df[[' no_of_dependents', ' income_annum', ' loan_amount', ' loan_term', ' cibil_score', ' residential_assets_value', ' luxury_assets_value']]
y = df[" loan_status"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
confusion_matrix(y_test,y_pred)
```

```python
x1 = df[[' no_of_dependents', ' income_annum', ' loan_amount', ' loan_term', ' cibil_score', ' residential_assets_value', ' luxury_assets_value']]
y1 = df[" loan_status"]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x1_scaled = scaler.fit_transform(x1)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x1_scaled,y1,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model1 = LogisticRegression()
model1.fit(x_train,y_train)
```

```python
y1_pred = model1.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
accuracy_score(y_test,y1_pred)
```

```python
print(classification_report(y_test,y1_pred))
```

```python
confusion_matrix(y_test,y1_pred)
```

```python
import pickle

# Save the trained model to a file
with open('loan_model.pkl', 'wb') as model_file:
    pickle.dump(model1, model_file)

# Save the scaler to a file (crucial for scaling new user inputs during deployment)
with open('scaler.pkl', 'wb') as scaler_file:
    pickle.dump(scaler, scaler_file)

print("Files successfully saved!")
print("- loan_model.pkl (Trained Logistic Regression Model)")
print("- scaler.pkl (Feature Scaler)")
```
